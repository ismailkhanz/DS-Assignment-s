Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
limit = 3
count = 0
for user in users:
    if count == limit:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
    count += 1
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28] 
critical = 40
for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
    print(f"Temperature: {temp}")
    
```

# Question 4

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True} ]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": True},{"name": "Sara", "paid": False},{"name": "John", "paid": True} ]
for customer in customers:
    if not customer['paid']:
        continue
    print(f"Access granted to: {customer["name"]}")
```

# Question 6

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']

```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains=[]
for email in emails:
    domain=email.split("@")[1]
    domains.append(domain)
print(f"domains: {domains}")
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4] multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2 2 x 2 = 4 2 x 3 = 6 3 x 1 = 3 3 x 2 = 6 3 x 3 = 9 4 x 1 = 4 4 x 2 = 8 4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]
for number in numbers:
    for multiplier in multipliers:
        print(f"{number} X {multiplier} = {number * multiplier}")
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc Incorrect password. Try again.

Enter password: 12345 Incorrect password. Try again.

Enter password: python123 Access granted!

Use a while loop.

```python
correct_password = "python123"
password = input("Enter password: ")
while password != correct_password:
    print(f"Incorrect password. Try again.")
    password = input("Enter password: ")
print("Access granted!")
```

Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5 4 3 2 1 Time's up!

Use a while loop.

```python
count = 5
while count >= 1:
    print(count, end=" ")
    count -= 1
print("Time's up!")
```

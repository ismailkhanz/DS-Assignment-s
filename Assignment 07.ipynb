Make a class called Employee. The init() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri- butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each instance. Then do some updations with it's methods.

```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("-" * 40)
        print(f"Name               : {self.name}")
        print(f"Employee Number    : {self.number}")
        print(f"Address            : {self.address}")
        print(f"Salary             : {self.salary:,.2f}")
        print(f"Post               : {self.post}")
        print(f"Year of Joining    : {self.year_of_joining}")
        print(f"Education          : {self.education}")
        print(f"Years of Experience: {self.year_of_experience}")
        print("-" * 40)

    def update_year_of_experience(self, years=1):
        self.year_of_experience += years
        print(f"{self.name}'s experience updated to "
              f"{self.year_of_experience} year(s).")

    def increase_salary(self, percent=10):
        increment = self.salary * percent / 100
        self.salary += increment
        print(f"{self.name}'s salary increased by {percent}% "
              f"(+{increment:,.2f}). New salary: {self.salary:,.2f}")
employee101 = Employee("Ali Khan", 101, "Gulshan-e-Iqbal, Karachi", 80000,
                       "Software Engineer", 2020, "BS Computer Science", 5)

print("=== employee101: individual attributes ===")
print("Name   :", employee101.name)
print("Number :", employee101.number)
print("Post   :", employee101.post)
print("Salary :", employee101.salary)

print("\n=== employee101: calling all methods ===")
employee101.describe_employee()
employee101.increase_salary(10)
employee101.update_year_of_experience()
employee101.describe_employee()

employees = [
    Employee("Ahmed Raza", 102, "DHA, Karachi", 95000, "Project Manager", 2016, "MBA", 9),
    Employee("Sara Ahmed", 103, "Clifton, Karachi", 70000, "HR Officer", 2019, "BBA", 6),
    Employee("Bilal Hussain", 104, "North Nazimabad, Karachi", 60000, "Accountant", 2021, "B.Com", 4),
    Employee("Ayesha Siddiqui", 105, "Gulistan-e-Johar, Karachi", 85000, "Data Analyst", 2018, "MS Statistics", 7),
    Employee("Usman Tariq", 106, "Saddar, Karachi", 55000, "Network Technician", 2022, "DAE Electronics", 3),
    Employee("Fatima Noor", 107, "PECHS, Karachi", 110000, "Senior Developer", 2014, "MS Software Engineering", 11),
    Employee("Hamza Sheikh", 108, "Malir, Karachi", 50000, "Sales Executive", 2023, "BBA", 2),
    Employee("Zainab Malik", 109, "Korangi, Karachi", 75000, "Graphic Designer", 2020, "BS Fine Arts", 5),
    Employee("Imran Qureshi", 110, "Bahadurabad, Karachi", 65000, "Office Manager", 2017, "BA", 8),
    Employee("Mariam Baig", 111, "Federal B Area, Karachi", 90000, "QA Lead", 2017, "BS Computer Science", 8),
]

print("\n=== Describing all 10 employees ===")
for emp in employees:
    emp.describe_employee()
print("\n=== Applying updates ===")
employees[0].increase_salary(15)
employees[1].increase_salary()            # default 10%
employees[2].update_year_of_experience()  # default +1 year
employees[3].update_year_of_experience(2)
employees[4].increase_salary(5)
employees[4].update_year_of_experience()
employees[6].increase_salary(20)
employees[9].update_year_of_experience()
employees[9].increase_salary(12)

print("\n=== Updated records ===")
for index in (0, 1, 2, 3, 4, 6, 9):
    employees[index].describe_employee()
```
